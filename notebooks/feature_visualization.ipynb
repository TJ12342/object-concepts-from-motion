# Motion Object Encoder: PCA Feature Visualization

This notebook runs a forward pass of the 421M-frame Swin-H model and visualizes its dense features with per-image PCA.

```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt

OPEN_SOURCE = Path.cwd().resolve()
if not (OPEN_SOURCE / 'tools').is_dir():
    OPEN_SOURCE = OPEN_SOURCE.parent
sys.path.insert(0, str(OPEN_SOURCE / 'tools'))
from feature_visualization import load_model, show_feature_map
```

```python
device = 'cuda' if __import__('torch').cuda.is_available() else 'cpu'
model = load_model(device=device)
```

```python
for image_name in ['kitti_highway.jpg', 'kitti_multicar.jpg']:
    print(image_name)
    show_feature_map(model, OPEN_SOURCE / 'assets' / image_name, device=device)
    plt.show()
```
